# Selecting Rows and Columns in Pandas

## Purpose

This notebook demonstrates selecting rows and columns using indexing and slicing in Pandas DataFrames.

```python
import pandas as pd
```

```python
data_frame = pd.read_csv("../data/raw/sample_data.csv")

print(data_frame)
```

## Selecting a Single Column

We can select a single column using its column name.

```python
print(data_frame["Name"])
```

## Selecting Multiple Columns

We can select multiple columns using a list of column names.

```python
print(data_frame[["Name", "City"]])
```

## Selecting Rows by Position

We use iloc for position-based indexing.

```python
print(data_frame.iloc[0])
print(data_frame.iloc[1:3])
```

## Selecting Rows by Label

We use loc for label-based indexing.

```python
print(data_frame.loc[0])
print(data_frame.loc[1:2])
```

## Selecting Rows and Columns Together

We can combine row and column selection together.

```python
print(data_frame.loc[0:2, ["Name", "Age"]])
```

## Conclusion

- Columns can be selected by name
- iloc uses positional indexing
- loc uses label-based indexing
- Combined selection helps extract specific subsets of data
