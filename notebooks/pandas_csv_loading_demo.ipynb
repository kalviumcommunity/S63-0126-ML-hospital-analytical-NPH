# Loading CSV Data into Pandas DataFrame

## Purpose

This notebook demonstrates how to load a CSV file into a Pandas DataFrame and inspect its structure.

```python
import pandas as pd
```

## Loading the CSV File

The following code loads the CSV file into a Pandas DataFrame.

```python
data_frame = pd.read_csv("../data/raw/sample_data.csv")

print("CSV Loaded Successfully")
```

## Previewing the Data

We use head() to inspect the first few rows of the dataset.

```python
print(data_frame.head())
```

## Inspecting Rows and Columns

This helps verify the structure of the DataFrame.

```python
print("Shape:", data_frame.shape)
print("Columns:", data_frame.columns)
```

## Conclusion

- Pandas loads CSV data into DataFrames
