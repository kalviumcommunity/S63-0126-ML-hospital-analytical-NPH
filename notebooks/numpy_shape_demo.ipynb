# NumPy Array Shape, Dimensions, and Indexing

```python
import numpy as np
```

## 1D Array Example

This example demonstrates a one-dimensional NumPy array along with its shape and dimensions.

```python
one_d_array = np.array([10, 20, 30, 40])

print("1D Array:")
print(one_d_array)

print("Shape:", one_d_array.shape)
print("Dimensions:", one_d_array.ndim)
```

## Accessing Elements in 1D Array

NumPy uses zero-based indexing, meaning indexing starts at 0.

```python
print("First element:", one_d_array[0])
print("Third element:", one_d_array[2])
```

## 2D Array Example

This example demonstrates a two-dimensional NumPy array with rows and columns.

```python
two_d_array = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("2D Array:")
print(two_d_array)

print("Shape:", two_d_array.shape)
print("Dimensions:", two_d_array.ndim)
```

## Accessing Elements in 2D Array

In 2D arrays, indexing uses row position first and column position second.

```python
print("Element at row 0, column 1:", two_d_array[0, 1])
print("Element at row 1, column 2:", two_d_array[1, 2])
```

## Conclusion

- Shape describes how data is organized
- ndim shows the number of dimensions
- Indexing starts at 0
- Rows come before columns in 2D indexing
