# Kernel Management Demonstration

This notebook demonstrates controlling the Jupyter kernel, including execution order, interrupting a long-running cell, and restarting the kernel.

## 1. Execution Order & State
Run the next two cells normally to establish some state in the kernel.

```python
message = "The kernel is running and storing variables!"
counter = 0
print(message)
```

```python
counter += 1
print(f"Counter is now at: {counter}")
```

## 2. Interrupting Execution
Run the following cell. It runs for 10 seconds. While it is running (indicated by `In [*]`), click the **Interrupt** button (the stop square in the toolbar) or use `Kernel -> Interrupt`.

```python
import time

print("Starting a 10-second process. Please interrupt me!")
try:
    for i in range(10):
        time.sleep(1)
        print(f"Working... {i+1}/10")
    print("\nProcess finished naturally.")
except KeyboardInterrupt:
    print("\nSuccess! You interrupted the kernel safely.")
```

Notice that interrupting the kernel stops the current cell but keeps the kernel alive. You can verify the variables are still intact by running the cell below:

```python
print(f"Message is still: {message}")
print(f"Counter is still: {counter}")
```

## 3. Restarting the Kernel
Now, let's restart the kernel. Click `Kernel -> Restart Kernel...`. This will clear all variables from memory.

After restarting, try running the cell below without running anything else first. It should result in a `NameError` because `message` is no longer defined.

```python
try:
    print(message)
except NameError as e:
    print(f"Error caught: {e}")
    print("This is expected! The kernel restart cleared the variables.")
```

## 4. Run All (Clean State)
Finally, select `Kernel -> Restart & Run All` (or `Restart Kernel and Run All Cells`). Watch as all cells execute from the top down in a clean environment, resulting in sequential execution counts `[1]`, `[2]`, etc. (The loop cell will take 10 seconds to finish before executing the rest).
