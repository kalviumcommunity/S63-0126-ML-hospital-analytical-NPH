# Python For and While Loops

```python
# For Loop Example

for i in range(1, 5):
    print("Iteration:", i)
```

```python
# Loop over list

numbers = [10, 20, 30]

for num in numbers:
    print("Value:", num)
```

```python
# While Loop Example

count = 1

while count <= 5:
    print("Count:", count)
    count += 1
```

```python
# Loop Control

for i in range(1, 6):
    if i == 3:
        continue   # skip 3
    if i == 5:
        break      # stop loop at 5
    print("Number:", i)
```
