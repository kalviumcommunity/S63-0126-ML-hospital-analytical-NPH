# NumPy Broadcasting Demonstration

```python
import numpy as np
```

## Scalar Broadcasting Example

A scalar value can be applied to every element of a NumPy array automatically.

```python
array_1d = np.array([1, 2, 3, 4])

print("Original Array:")
print(array_1d)

print("Shape:", array_1d.shape)

result_scalar = array_1d + 10

print("After Broadcasting:")
print(result_scalar)
```

## 1D to 2D Broadcasting Example

A 1D array can be broadcast across rows of a 2D array when shapes are compatible.

```python
array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

array_1d = np.array([10, 20, 30])

print("2D Array Shape:", array_2d.shape)
print("1D Array Shape:", array_1d.shape)

result_broadcast = array_2d + array_1d

print("Broadcasted Result:")
print(result_broadcast)
```

## Conclusion

- Broadcasting allows operations on arrays with different shapes
- NumPy automatically aligns compatible dimensions
- Shape awareness helps prevent broadcasting errors
