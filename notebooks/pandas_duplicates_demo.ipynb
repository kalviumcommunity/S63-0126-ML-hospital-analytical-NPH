# Handling Duplicate Records in Pandas

## Purpose

This notebook demonstrates how to detect, inspect, and remove duplicate records in a Pandas DataFrame.

```python
import pandas as pd
```

```python
data_frame = pd.read_csv("../data/raw/duplicate_data.csv")

print("Original Data:")
print(data_frame)
```

## Detecting Duplicate Rows

```python
duplicates = data_frame.duplicated()

print("Duplicate Rows:")
print(duplicates)

print("Total Duplicates:", duplicates.sum())
```

## Inspecting Duplicate Entries

```python
print("Duplicate Records:")
print(data_frame[data_frame.duplicated()])
```

## Removing Duplicate Rows

```python
data_frame_cleaned = data_frame.drop_duplicates()

print("Data After Removing Duplicates:")
print(data_frame_cleaned)
```

## Comparing Shapes Before and After

```python
print("Original Shape:", data_frame.shape)
print("New Shape:", data_frame_cleaned.shape)
```

## Verification

Duplicates have been removed successfully.
