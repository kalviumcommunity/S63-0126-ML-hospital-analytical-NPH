# Python Numeric and String Data Types


```python
# Numeric Data Types

a = 10          # integer
b = 5.5         # float

print("Integer:", a)
print("Float:", b)

print("Addition:", a + b)
print("Multiplication:", a * b)
print("Division:", a / 2)
```

```python
# String Data Types

name = "Hasan"
greeting = "Hello"

print(greeting + " " + name)
```

```python
# Mixing Numbers and Strings

age = 21

# Wrong (will cause error if uncommented)
# print("Age is " + age)

# Correct way
print("Age is " + str(age))
```

```python
# Checking Data Types

x = 10
y = "Python"

print(type(x))
print(type(y))
```
